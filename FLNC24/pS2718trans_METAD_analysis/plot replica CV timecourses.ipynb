```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.ndimage
from scipy.ndimage import gaussian_filter
path = '/Volumes/Elements/PTM_project/FLNC24/pS2718trans_METAD/DATA_PLUMED/'
```

```python

nreplica=3


plt.figure(figsize=(8,3))

for i in range(nreplica):
    
    plt.subplot(1,3,i+1)
    
    dat = np.loadtxt(path+'COLVAR.{}'.format(i+1))
    
    plt.title('SIM{}'.format(i+1))
    plt.scatter(dat[:,0]/1000,dat[:,1],alpha = 0.05,s=0.2,color = 'black')
    
    plt.ylim(-np.pi,np.pi)
    
    plt.xlabel('Time (ns)')
    plt.ylabel('Zeta')
    
plt.subplots_adjust(hspace = 0.2,wspace = 0.2)
plt.tight_layout()
plt.savefig('replicas_zeta.png')
```
