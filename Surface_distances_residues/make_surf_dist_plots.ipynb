```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import MDAnalysis.analysis.distances
import MDAnalysis as md
import mdtraj as mdj
import MDAnalysis.analysis.contacts
import pandas as pd
import MDAnalysis.analysis.rms
import os
from MDAnalysis.analysis.dihedrals import Dihedral
from MDAnalysis.analysis.rms import RMSF
from MDAnalysis.analysis import rms, align
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42

path = '/Volumes/Elements/PTM_project/Surface_distances_analysis/'
```

```python
# individual simulation plots of the buried PTMs

protein = 'ACYP1'
res = 'SER71'
nsims=3
Nmov = 50

plt.figure(figsize= (4,5))

for s in range(1,nsims+1):
    
    plt.subplot(3,1,s)
    
    
    dat = np.load(path+"{}_sim{}_{}.npy".format(protein,s,res))
    dat = dat[::5] # downsample 10k, 0.1 ns  frequency

    
    time = np.arange(dat.shape[0])/10 # ns
    
    mov_x = np.convolve(time, np.ones(Nmov)/Nmov, mode='valid')
    mov_y = np.convolve(dat, np.ones(Nmov)/Nmov, mode='valid')

    plt.plot(time,dat,color = '#FEC89A',linewidth = 0.5, rasterized=True)
    plt.plot(mov_x,mov_y,color = '#F49595',linewidth = 1.5, rasterized=True)
    
    plt.ylim(0,1)
    plt.xlim(0,1000)

    if s==1:
        plt.title(protein+str(' ')+res)
    elif s==2:
        plt.ylabel('Surfance distance (nm)')
    elif s==3:
        plt.xlabel('Time (ns)')

plt.tight_layout()
plt.savefig('Surdist_timecourses_{}_{}.pdf'.format(protein,res), rasterized=True)
```
