```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.ndimage
from scipy.ndimage import gaussian_filter
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42



path = '/Volumes/Elements/PTM_project/FLN5_proline_METAD/'
```

```python
system = 'FLN5_FL'
sim = 3
zeta = np.loadtxt(path+system+'/final_bias_from_hills_1us/COLVAR_REW.{}'.format(sim))[:,1]
time = np.loadtxt(path+system+'/sim{}/sasaMET741.xvg'.format(sim),skiprows=25)[:,0]/1000
time = time[::25]
zeta = zeta[::25]
SASA = SASA[::25]
print(zeta.shape)

fontsize = 10


plt.figure(figsize = (4.5,2.))
sc = plt.scatter(time,zeta,color='black',s=3, rasterized=True)
plt.xlabel('Time (ns)',fontsize = fontsize)
plt.ylabel('\u03B6',fontsize = fontsize)
plt.ylim(-np.pi,np.pi)
plt.yticks(ticks=[-np.pi,0,np.pi],labels=['-\u03C0',0,'\u03C0'])
plt.xlim(0,1000)

plt.tight_layout()
plt.savefig('{}_ZETA_SIM{}.pdf'.format(system,sim), dpi=300)


```
