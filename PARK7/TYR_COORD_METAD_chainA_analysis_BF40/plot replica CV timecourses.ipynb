```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.ndimage
from scipy.ndimage import gaussian_filter
path = '/Volumes/Elements/PTM_project/PARK7/TYR_COORD_METAD_chainA_BF40/DATA_PLUMED/'
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42
```

```python

nreplica=3


plt.figure(figsize=(8,3))

for i in range(nreplica):
    
    plt.subplot(1,3,i+1)
    
    dat = np.loadtxt(path+'COLVAR.{}'.format(i+1))
    
    plt.title('SIM{}'.format(i+1))
    plt.scatter(dat[:,0]/1000,dat[:,1],alpha = 0.05,s=0.2,color = 'black')
    
    plt.ylim(-np.pi,np.pi)
    
    plt.xlabel('Time (ns)')
    plt.ylabel('Zeta')
    
plt.subplots_adjust(hspace = 0.2,wspace = 0.2)
plt.tight_layout()
plt.savefig('replicas_zeta.png')
```

```python

ymax = 8

nsims=3

plt.figure(figsize = (4,2))

colors = ['#F49595','#F1CEBE','#F8EDEB']

vals = []

for i in range(1,nsims+1):
    
    dat = np.loadtxt(path+'../RMSD_SASA_data/rmsd_BB_{}.xvg'.format(i),skiprows=18)[::200]
    
    rmsd = dat[:,1]*10 # convert to Angstrom
    time = dat[:,0]/1000 # convert to nanoseconds
    
    vals.append(np.mean(rmsd)) # mean 

    plt.plot(time,rmsd,color = colors[i-1],alpha=0.9,linewidth = 1,label = 'Sim. {}'.format(i))
    
plt.legend(ncol=3)
print('{:.1f} ± {:.1f} $\\AA$'.format(np.mean(vals),np.std(vals)/np.sqrt(nsims)))
plt.ylim(0,ymax)
plt.xlim(0,np.max(time))

    
plt.xlabel('Time (ns)')
plt.ylabel('BB-RMSD ($\\AA$)')

plt.tight_layout()
plt.savefig('BB_RMSD_METAD.pdf')





```
