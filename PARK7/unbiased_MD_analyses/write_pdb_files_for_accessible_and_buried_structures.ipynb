```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import MDAnalysis.analysis.distances
import MDAnalysis as md
import mdtraj as mdj
import MDAnalysis.analysis.contacts
import pandas as pd
import MDAnalysis.analysis.rms
import os
from MDAnalysis.analysis.dihedrals import Dihedral
from MDAnalysis.analysis.rms import RMSF
from MDAnalysis.analysis import rms, align
import matplotlib
matplotlib.rcParams['pdf.fonttype'] = 42

path = '/Volumes/Elements/PTM_project/PARK7/unbiased_1us/'
```

```python
sim=1

u = md.Universe(path+'sim{}/processed.pdb'.format(sim),path+'sim{}/fit.xtc'.format(sim))
protein = u.select_atoms('all')

resids=[47,57]
restypes=['SER','SER']

for resid,resname in zip(resids,restypes):
    
    print('Analysing residue {}{}...'.format(resname,resid))
    
    sasaA=np.loadtxt(path+"sim{}/sasa_chainA_{}{}.xvg".format(sim,resname,resid),skiprows=25)
    sasaB=np.loadtxt(path+"sim{}/sasa_chainB_{}{}.xvg".format(sim,resname,resid),skiprows=25)
    sasa=np.mean([sasaA[:,2],sasaB[:,2]],axis=0) # average SASA of both chains
    
    # extract second half of trajectory (ensuring structure/system is well equilibrated at that point)
    sasa_secondhalf=sasa[int(sasa.shape[0]/2):]
    # find minimum and maximum
    sasa_min = np.min(sasa_secondhalf)
    sasa_max = np.max(sasa_secondhalf)

    # find first frames where min/max values are
    idx_min = np.where(sasa_secondhalf==sasa_min)[0][0]+int(sasa.shape[0]/2) # the +int(sasa.shape[0]/2) shifts the index to the second half
    idx_max = np.where(sasa_secondhalf==sasa_max)[0][0]+int(sasa.shape[0]/2)

    time_min = sasaA[:,0][idx_min]
    time_max = sasaA[:,0][idx_max]

    print('Minimum SASA: {:.2f} nm^2 at {:.1f} ps'.format(sasa_min,time_min))
    print('Maximum SASA: {:.2f} nm^2 at {:.1f} ps'.format(sasa_max,time_max))

    # write pdb structures
    u.trajectory[idx_min]
    with md.Writer(path+'sim{}/minSASA_RESID{}_{:.2f}nm2_at_{:.1f}ps.pdb'.format(sim,resid,sasa_min,time_min),protein.n_atoms) as W:
        W.write(protein)

    u.trajectory[idx_max]
    with md.Writer(path+'sim{}/maxSASA_RESID{}_{:.2f}nm2_at_{:.1f}ps.pdb'.format(sim,resid,sasa_max,time_max),protein.n_atoms) as W:
        W.write(protein)

    print('Structures written to PDB files!')
```
